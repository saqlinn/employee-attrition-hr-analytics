```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:\\Users\\saqli\\Downloads\\workspace\\Employee-Attrition-Project\\data\\employee_attrition.csv")

df.head()
```

```python
# Create a copy
clean_df = df.copy()

# 1. Data quality assessment
print("Shape:", clean_df.shape)
print("\nMissing values:")
print(clean_df.isnull().sum())

print("\nDuplicate rows:")
print(clean_df.duplicated().sum())

print("\nData types:")
print(clean_df.dtypes)

# 2. Missing value check
missing_percentage = (
    clean_df.isnull().sum() / len(clean_df)
) * 100

print("\nMissing percentage:")
print(missing_percentage)

# 3. Remove duplicates
clean_df = clean_df.drop_duplicates()

# 4. Reset index
clean_df = clean_df.reset_index(drop=True)

# 5. Consistency checks
print("\nUnique Attrition values:")
print(clean_df["Attrition"].unique())

print("\nUnique Feedback values:")
print(clean_df["AverageFeedback"].unique())

print("\nUnique Pay Increment values:")
print(clean_df["LastPayIncrementBand"].unique())

print("\nUnique Promotion values:")
print(clean_df["LastPromotionYears"].unique())

# 6. Validation
print("\nFinal Shape:", clean_df.shape)

print("\nFinal Missing Values:")
print(clean_df.isnull().sum())

print("\nFinal Duplicate Count:")
print(clean_df.duplicated().sum())

print("\nFinal Statistics:")
print(clean_df.describe())

# 7. Save clean dataset
clean_df.to_csv(
    "../data/employee_attrition_clean.csv",
    index=False
)

print("\nClean dataset saved successfully.")
```
