# Employee Statistics Report

## 1. Objective

The objective of this analysis is to use NumPy to perform numerical and statistical analysis on the Employee Attrition dataset.

## 2. Dataset Overview

The dataset contains 1,000 employee records and 8 columns. The dataset contains employee experience, organizational tenure, pay increment, feedback, promotion-related information, and a binary attrition target.

## 3. Employee Experience Statistics

The TotalMonthsOfExp feature was analyzed using NumPy functions such as mean, median, minimum, maximum, standard deviation, and sum.

## 4. Organization and Tenure Analysis

TotalOrgsWorked and MonthsInOrg were analyzed to understand employee mobility and organizational tenure.

## 5. Compensation and Feedback Analysis

LastPayIncrementBand and AverageFeedback were analyzed using statistical measures such as mean, median, minimum, and maximum.

## 6. Promotion Analysis

LastPromotionYears was analyzed using NumPy statistical functions to understand its distribution.

## 7. Attrition Analysis

The Attrition variable contains two classes, 0 and 1. The dataset contains 770 records in class 0 and 230 records in class 1, corresponding to 77% and 23% respectively.

## 8. Key Observations

- The dataset contains 1,000 employee records.
- TotalMonthsOfExp provides information about employee professional experience.
- TotalOrgsWorked provides information about employee organizational mobility.
- MonthsInOrg provides information about organizational tenure.
- LastPayIncrementBand and AverageFeedback provide additional employee-related information.
- Attrition is a binary target variable with two classes.
- The target classes are not equally distributed.
- NumPy was used for numerical calculations, statistical analysis, filtering, and aggregation.

## 9. Conclusion

NumPy provides efficient tools for numerical and statistical analysis.
