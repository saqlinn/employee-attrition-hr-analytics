```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("C:\\Users\\saqli\\Downloads\\workspace\\Employee-Attrition-Project\\data\\employee_attrition.csv")
```

```python
df.head()
```

```python
sns.countplot(x="Attrition", data=df)

plt.title("Employee Distribution by Attrition")
plt.xlabel("Attrition")
plt.ylabel("No.of Employees")
```

```python
plt.figure(figsize=(8,6))

plt.hist(df['TotalMonthsOfExp'], bins=20)
plt.title("Distribution of Total Months of Experience")
plt.xlabel("Total Months of Experience")
plt.ylabel("No.of Employees")
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="TotalMonthsOfExp",
    y="MonthsInOrg",
    hue="Attrition",
    data=df
)

plt.title("Experience vs Months in Organization")
plt.xlabel("Total Experience (Months)")
plt.ylabel("Months in Organization")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["TotalMonthsOfExp"])

plt.title("Experience Box Plot")
plt.xlabel("Total Experience (Months)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Attrition",
    y="TotalMonthsOfExp",
    data=df
)

plt.title("Experience Distribution by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Total Experience (Months)")

plt.show()
```

```python
plt.figure(figsize=(10, 7))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Employee Feature Correlation Heatmap")

plt.show()
```
