# Data Dictionary

| Column | Data Type | Role | Description |
|---|---|---|---|
| EmployeeID | Integer / Identifier | Identifier | Unique identifier assigned to each employee |
| TotalMonthsOfExp | Numerical | Feature | Total professional experience of the employee in months |
| TotalOrgsWorked | Discrete Numerical | Feature | Number of organizations the employee has worked for |
| MonthsInOrg | Numerical | Feature | Number of months the employee has spent in the organization |
| LastPayIncrementBand | Ordinal | Feature | Employee's latest pay increment band |
| AverageFeedback | Ordinal | Feature | Average feedback/rating level of the employee |
| LastPromotionYears | Discrete Numerical / Ordinal | Feature | Number of years associated with the employee's last promotion |
| Attrition | Binary | Target | Binary outcome representing employee attrition |

# Feature Importance Discussion

## TotalMonthsOfExp
Total professional experience may be related to employee career stage and mobility patterns.

## TotalOrgsWorked
The number of organizations previously worked at may provide information about an employee's job mobility.

## MonthsInOrg
The duration an employee has spent in the current organization may help analyze tenure-related attrition patterns.

## LastPayIncrementBand
Pay increment information may provide insight into compensation-related employee patterns.

## AverageFeedback
Feedback or performance-related information may help analyze employee performance patterns in relation to attrition.

## LastPromotionYears
Time associated with the employee's last promotion may provide information about career progression.

## EmployeeID
EmployeeID is an identifier and does not represent an employee characteristic. Therefore, it should generally not be used as a predictive feature.

# Business Interpretation

The dataset contains employee experience, organizational tenure, pay progression, feedback, promotion-related information and attrition outcomes.

These attributes can be used to investigate employee workforce patterns and identify characteristics associated with attrition.

From an HR analytics perspective:

- Experience helps understand employee career stage.
- Number of organizations worked at provides information about employee mobility.
- Months in the organization represents employee tenure.
- Pay increment band provides information about compensation progression.
- Average feedback provides information about employee feedback/performance.
- Promotion-related information provides information about career progression.
- Attrition represents the outcome that the machine learning model will attempt to predict.

# Dataset Summary

The dataset contains 1,000 employee records and 8 columns.

The dataset is structured data represented in tabular form. It contains an employee identifier, employee experience information, organizational tenure, pay increment information, feedback, promotion-related information and an attrition target.

The Attrition variable contains two classes, 0 and 1. Class 0 represents 77% of the records, while class 1 represents 23% of the records.

The dataset contains numerical, discrete numerical, ordinal and identifier variables. EmployeeID is treated as an identifier rather than a predictive feature.

The dataset will be used in subsequent stages for exploratory data analysis, preprocessing, feature engineering and machine learning.
