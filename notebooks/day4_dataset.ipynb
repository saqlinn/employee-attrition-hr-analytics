## 1. Objective

The objective of this analysis is to use Pandas for data inspection, filtering, sorting, grouping, aggregation, and basic reporting on the Employee Attrition dataset.

## 2. Dataset Overview

The dataset contains 1,000 employee records and 8 columns. The attributes include employee experience, number of organizations worked for, organizational tenure, pay increment band, average feedback, promotion-related information, and attrition.

## 3. Group-wise Analysis

Employees were grouped based on the Attrition variable. Pandas GroupBy and aggregation functions were used to calculate descriptive statistics for experience, organizational tenure, feedback, and other employee attributes.

## 4. Aggregation

The following statistics were calculated for each attrition class:

- Average and median total experience
- Average number of organizations worked for
- Average and median months in organization
- Average feedback
- Average promotion-related value

## 5. Key Observations

- The dataset contains two attrition classes.
- The classes have different employee counts.
- Group-wise statistics allow numerical employee characteristics to be compared between the two classes.
- Pandas GroupBy provides an efficient way to perform such grouped analysis.

## 6. Conclusion

Pandas was used to inspect, filter, sort, group, aggregate, and summarize the employee dataset. The analysis provides a foundation for the exploratory data analysis stage of the Employee Attrition Prediction project.
